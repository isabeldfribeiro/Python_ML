```python
#Importando biblioteca
import pandas as pd
```

```python
#Chamando df
df = pd.read_csv("../tabelas/tabela_estruturada.csv")
df
```

```python
# Variação com base no grupo

colunas = [
    col for col in df.columns
    if col != "grupo"
]

resultado = {}

for grupo in df["grupo"].dropna().unique():

    df_grupo = df[df["grupo"] == grupo]

    for coluna in colunas:

        counts = df_grupo[coluna].value_counts(dropna=True)
        total = counts.sum()

        resultado.setdefault(coluna, []).append(
            " | ".join(
                f"{variavel} - {n} - {(n / total * 100):.2f}%"
                for variavel, n in counts.items()
            )
        )

var_grupo = pd.DataFrame(
    resultado,
    index=df["grupo"].dropna().unique()
)

var_grupo.index.name = "grupo"

var_grupo.to_csv("../tabelas/tabela_grupo.csv")

var_grupo
```

```python
# Variação com base na MIC

colunas = [
    col for col in df.columns
    if col != "MIC_imipenem"
]

resultado = {}

for mic in df["MIC_imipenem"].dropna().unique():

    df_mic = df[df["MIC_imipenem"] == mic]

    for coluna in colunas:

        counts = df_mic[coluna].value_counts(dropna=True)
        total = counts.sum()

        resultado.setdefault(coluna, []).append(
            " | ".join(
                f"{variavel} - {n} - {(n / total * 100):.2f}%"
                for variavel, n in counts.items()
            )
        )

var_MIC = pd.DataFrame(
    resultado,
    index=df["MIC_imipenem"].dropna().unique()
)

var_MIC.index.name = "MIC_imipenem"

var_MIC.to_csv("../tabelas/tabela_MIC.csv")

var_MIC
```

```python
# Variação das variáveis em relação à MIC_imipenem dentro de cada grupo

colunas = [
    col for col in df.columns
    if col not in ["grupo", "MIC_imipenem"]
]

resultado = []

for (grupo, mic_imipenem), df_sub in df.dropna(subset=["grupo", "MIC_imipenem"]).groupby(
    ["grupo", "MIC_imipenem"],
    sort=True
):

    linha = {
        "grupo": grupo,
        "MIC_imipenem": mic_imipenem
    }

    for coluna in colunas:

        counts = df_sub[coluna].value_counts(dropna=True)
        total = counts.sum()

        linha[coluna] = " | ".join(
            f"{variavel} - {n} - {(n / total * 100):.2f}%"
            for variavel, n in counts.items()
        )

    resultado.append(linha)

var_grupo_MIC_imipenem = pd.DataFrame(resultado)

var_grupo_MIC_imipenem.to_csv(
    "../tabelas/grupo_MIC_imipenem.csv",
    index=False
)

var_grupo_MIC_imipenem
```
